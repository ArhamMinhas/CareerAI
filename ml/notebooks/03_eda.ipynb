# 03 — EDA

Real distributions over the snapshot — job categories, salary spread, posting dates.

```python
from training.data import load_snapshot

jobs = load_snapshot("jobs")
print(jobs["search_category"].value_counts())
```

```python
salary_mid = (jobs["salary_min"] + jobs["salary_max"]) / 2
print(salary_mid.describe())
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
salary_mid.dropna().hist(bins=30, ax=ax)
ax.set_title("Real posted salary midpoints")
ax.set_xlabel("USD")
plt.show()
```

```python
posted = jobs["posted_at"].dropna()
fig, ax = plt.subplots()
posted.dt.to_period("W").value_counts().sort_index().plot(kind="bar", ax=ax)
ax.set_title("Real postings per week (posted_at)")
plt.show()
```
