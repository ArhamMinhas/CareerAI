# 08 — Model Evaluation

Reads every model's real, already-saved `metadata.json` (docs/ML_PIPELINE.md §6) and prints a single comparison table — the same honest summary in `ml/models/README.md`, generated from the actual artifacts rather than transcribed by hand.

```python
import json
from training.registry import MODELS_DIR

for model_dir in sorted(MODELS_DIR.iterdir()):
    if not model_dir.is_dir():
        continue
    versions = sorted(p for p in model_dir.iterdir() if p.is_dir())
    if not versions:
        continue
    metadata_path = versions[-1] / "metadata.json"
    if not metadata_path.exists():
        continue
    metadata = json.loads(metadata_path.read_text())
    print(f"{metadata['name']} ({metadata['version']}): {metadata['metric']}={metadata['score']:.4f}")
    print(f"  baseline: {metadata['baseline']}")
    print(f"  limitations: {metadata['limitations'][:200]}...")
    print()
```
