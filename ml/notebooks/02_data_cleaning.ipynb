# 02 — Data Cleaning

Real data quality check on the snapshot — nulls, duplicates, and the specific gaps this phase's ROADMAP entry documents (thin `job_skills`/`seniority_level` coverage).

```python
from training.data import load_snapshot

jobs = load_snapshot("jobs")
job_skills = load_snapshot("job_skills")
skills = load_snapshot("skills")

print("jobs:", jobs.shape)
print("jobs missing salary:", jobs["salary_min"].isna().sum())
print("jobs missing seniority_level:", jobs["seniority_level"].isna().sum(), "/", len(jobs))
print("jobs with a search_category:", jobs["search_category"].notna().sum(), "/", len(jobs))
print("job_skills rows:", len(job_skills), "-- distinct jobs covered:", job_skills["job_id"].nunique())
print("skills with an embedding:", skills["embedding"].notna().sum(), "/", len(skills))
print("duplicate job ids:", jobs["id"].duplicated().sum())
```
